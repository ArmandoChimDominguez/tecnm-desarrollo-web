## SECCIÓN A: TRACK PYTHON / FASTAPI
### Fase 1: El Artesano (15 min - Sin IA)
**Reto:** Anotación manual de metadata de endpoints

```python
%%writefile app_artesano.py
from fastapi import FastAPI, HTTPException, status
from pydantic import BaseModel


app = FastAPI(
    title="API Artesano - Sesión 07",
    description="API con anotaciones manuales para Swagger.",
    version="1.0.0"
)

class PayloadRequest(BaseModel):
    dato: str


@app.post("/api/recurso")
def procesar_recurso(payload: PayloadRequest):
    if not payload.dato.strip():
        raise HTTPException(status_code=status.HTTP_400_BAD_REQUEST, detail="Dato vacio")
    return {"status": "success", "mensaje": f"Procesado: {payload.dato}"}

# --- RETO DEL ARTESANO nueva ruta anotada manualmente
@app.get(
    "/api/saludo",
    tags=["Saludos"],
    summary="Devuelve un saludo básico",
    description="Este endpoint sirve para demostrar cómo se anota la metadata manualmente en FastAPI sin usar IA."
)
def obtener_saludo():
    return {"mensaje": "Hola desde la API Artesana"}
```

### Fase 2: El Científico (Prompting C-R-E-O)
Pega aquí la implementación optimizada propuesta por la IA:

```python
%%writefile app_ia.py
from fastapi import FastAPI

# Pega aquí el código generado por la IA para auditarlo (Aquí la IA usa json_schema_extra y una configuración de app mucho más robusta).
%%writefile app_ia.py
from fastapi import FastAPI
from pydantic import BaseModel, Field

# Metadata Avanzada (Documentación profesional)
app = FastAPI(
    title="API Científico - Sesión 07",
    description="API optimizada con documentación OpenAPI v3.1 interactiva.",
    version="2.1.0",
    contact={
        "name": "Soporte DAB-2201",
        "email": "soporte@tecnm.mx",
    }
)

# Modelo optimizado con Ejemplos para Swagger
class PayloadRequest(BaseModel):
    dato: str = Field(..., description="Texto a procesar", examples=["Texto de prueba"])

    model_config = {
        "json_schema_extra": {
            "examples": [
                {"dato": "Ejemplo de JSON para Swagger"}
            ]
        }
    }

# --- RUTA DE COMPATIBILIDAD PARA LA PRUEBA DEL PROFESOR ---
@app.post("/api/recurso", tags=["Compatibilidad"])
def procesar_recurso(payload: PayloadRequest):
    return {"status": "success", "mensaje": f"Procesado: {payload.dato}"}

# --- RETO DEL CIENTÍFICO: Endpoint bien documentado ---
@app.get(
    "/api/libros/{libro_id}",
    tags=["Libros"],
    summary="Obtiene un libro por su ID",
    description="Retorna los detalles de un libro. Si el ID no existe, lanza un 404.",
    response_description="Los datos del libro solicitado."
)
def obtener_libro(libro_id: int):
    if libro_id == 1:
        return {"id": 1, "titulo": "Cien Años de Soledad"}
    return {"id": libro_id, "titulo": "Libro de prueba"}
```

### Fase 3: El Auditor (Test Harness HTTP)
Pruebas de servidor: Verificación de renderizado de interfaz Swagger UI

```python
import threading
import time
import uvicorn
import httpx
from app_artesano import app

def run_server():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

server_thread = threading.Thread(target=run_server, daemon=True)
server_thread.start()
time.sleep(1.5)

client = httpx.Client(base_url="http://127.0.0.1:8000")
print("=== AUDITORIA CRITICA WEB: SESION 07 ===\n")
r = client.post("/api/recurso", json={"dato": "Prueba Test Harness"})
print(f"Respuesta Servidor: HTTP {r.status_code} - {r.json()}")
assert r.status_code == 200
print("PASO ✅ - Test Harness HTTP ejecutado con exito.")
```
