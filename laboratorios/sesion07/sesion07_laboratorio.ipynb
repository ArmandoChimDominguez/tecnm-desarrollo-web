### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI
from pydantic import BaseModel
from typing import List

# 1. Metadatos de la API
app = FastAPI(
    title="Biblioteca API",
    version="1.0.0",
    description="API de libros del curso DAB-2201"
)

# 2. Modelo Pydantic
class Libro(BaseModel):
    titulo: str
    anio: int
    autor: str

# 3. Endpoint GET (Devuelve el JSON exacto que pide el test)
@app.get("/api/libros")
def listar_libros():
    return {"libros": []}

# 4. Endpoint POST (Hace que el modelo Libro aparezca en Swagger)
@app.post("/api/libros", response_model=Libro)
def crear_libro(libro: Libro):
    return libro
````

````python
%%writefile app_ia.py
from fastapi import FastAPI, status
from pydantic import BaseModel, Field
from typing import List

app = FastAPI(
    title="Biblioteca API",
    version="1.0.0",
    description="API de libros del curso DAB-2201 con documentación optimizada"
)

# Modelo con descripciones detalladas
class Libro(BaseModel):
    titulo: str = Field(..., description="Título del libro", examples=["Rayuela"])
    anio: int = Field(..., description="Año de publicación", examples=[1963])
    autor: str = Field(..., description="Autor del libro", examples=["Julio Cortázar"])

# Modelo de respuesta para envolver la lista
class LibroListResponse(BaseModel):
    libros: List[Libro]

@app.get("/api/libros", response_model=LibroListResponse)
def listar_libros():
    return {"libros": []}

@app.post("/api/libros", response_model=Libro, status_code=status.HTTP_201_CREATED)
def crear_libro(libro: Libro):
    return libro
````
