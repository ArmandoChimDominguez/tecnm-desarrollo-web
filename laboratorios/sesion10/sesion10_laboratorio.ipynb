### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, HTTPException, status
from pydantic import BaseModel

app = FastAPI(title="Mock API Artesano - Hito 1")

class Libro(BaseModel):
    id: int
    titulo: str
    anio: int

class LibroRequest(BaseModel):
    titulo: str
    anio: int

db_libros = []
contador_id = 1

@app.get("/api/libros")
def listar_libros():
    return {"libros": db_libros}

@app.post("/api/libros", status_code=status.HTTP_201_CREATED, response_model=Libro)
def crear_libro(libro: LibroRequest):
    global contador_id

    if not libro.titulo.strip():
        raise HTTPException(status_code=422, detail="El título no puede estar vacío")
    if libro.anio < 1400 or libro.anio > 2100:
        raise HTTPException(status_code=422, detail="Año inválido")

    nuevo_libro = {
        "id": contador_id,
        "titulo": libro.titulo,
        "anio": libro.anio
    }
    db_libros.append(nuevo_libro)
    contador_id += 1
    return nuevo_libro

@app.get("/api/libros/{libro_id}", response_model=Libro)
def obtener_libro(libro_id: int):
    for libro in db_libros:
        if libro["id"] == libro_id:
            return libro
    raise HTTPException(status_code=404, detail="Libro no encontrado")

@app.put("/api/libros/{libro_id}", response_model=Libro)
def actualizar_libro(libro_id: int, libro: LibroRequest):
    if not libro.titulo.strip():
        raise HTTPException(status_code=422, detail="El título no puede estar vacío")

    for i, l in enumerate(db_libros):
        if l["id"] == libro_id:
            db_libros[i]["titulo"] = libro.titulo
            db_libros[i]["anio"] = libro.anio
            return db_libros[i]

    raise HTTPException(status_code=404, detail="Libro no encontrado")

@app.delete("/api/libros/{libro_id}", status_code=status.HTTP_204_NO_CONTENT)
def eliminar_libro(libro_id: int):
    for i, l in enumerate(db_libros):
        if l["id"] == libro_id:
            del db_libros[i]
            return
    raise HTTPException(status_code=404, detail="Libro no encontrado")
````

````python
%%writefile app_ia.py
from fastapi import FastAPI, HTTPException, status
from pydantic import BaseModel, Field

app = FastAPI(title="Mock API Científico - Hito 1")

class LibroRequest(BaseModel):
    titulo: str = Field(..., min_length=1, description="Título del libro")
    anio: int = Field(..., ge=1400, le=2100, description="Año de publicación")

# IMPORTANTE: El modelo se llama "Libro" para que aparezca en OpenAPI
class Libro(BaseModel):
    id: int
    titulo: str
    anio: int

db_libros = {}
contador_id = 1

@app.get("/api/libros")
def listar_libros():
    return {"libros": list(db_libros.values())}

@app.post("/api/libros", status_code=status.HTTP_201_CREATED, response_model=Libro)
def crear_libro(libro: LibroRequest):
    global contador_id
    nuevo_libro = {
        "id": contador_id,
        "titulo": libro.titulo,
        "anio": libro.anio
    }
    db_libros[contador_id] = nuevo_libro
    contador_id += 1
    return nuevo_libro

@app.get("/api/libros/{libro_id}", response_model=Libro)
def obtener_libro(libro_id: int):
    if libro_id not in db_libros:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    return db_libros[libro_id]

@app.put("/api/libros/{libro_id}", response_model=Libro)
def actualizar_libro(libro_id: int, libro: LibroRequest):
    if libro_id not in db_libros:
        raise HTTPException(status_code=404, detail="Libro no encontrado")

    db_libros[libro_id]["titulo"] = libro.titulo
    db_libros[libro_id]["anio"] = libro.anio
    return db_libros[libro_id]

@app.delete("/api/libros/{libro_id}", status_code=status.HTTP_204_NO_CONTENT)
def eliminar_libro(libro_id: int):
    if libro_id not in db_libros:
        raise HTTPException(status_code=404, detail="Libro no encontrado")
    del db_libros[libro_id]
````
