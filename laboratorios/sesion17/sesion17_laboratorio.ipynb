### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer
from pydantic import BaseModel
from jose import JWTError, jwt
import datetime

app = FastAPI(title="API Artesano - Sesión 17")

SECRET_KEY = "mi_clave_secreta"
ALGORITHM = "HS256"
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="/api/token")

# Usuarios exactos que pide el test
usuarios_db = {
    "admin": {"password": "admin123", "rol": "admin"},
    "ana": {"password": "secreta", "rol": "user"}
}

class LoginRequest(BaseModel):
    usuario: str
    password: str

# 1. Ruta de Token (/api/token)
@app.post("/api/token")
def login(datos: LoginRequest):
    if datos.usuario in usuarios_db and usuarios_db[datos.usuario]["password"] == datos.password:
        rol = usuarios_db[datos.usuario]["rol"]
        expiracion = datetime.datetime.utcnow() + datetime.timedelta(minutes=30)
        token = jwt.encode({"sub": datos.usuario, "rol": rol, "exp": expiracion}, SECRET_KEY, algorithm=ALGORITHM)
        return {"access_token": token, "token_type": "bearer"}

    raise HTTPException(status_code=401, detail="Credenciales incorrectas")

# 2. Ruta de Perfil (/api/perfil)
@app.get("/api/perfil")
def perfil(token: str = Depends(oauth2_scheme)):
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        return {"usuario": payload.get("sub"), "rol": payload.get("rol")}
    except JWTError:
        raise HTTPException(status_code=401, detail="Token inválido")

# 3. Ruta de Admin (/api/admin/panel)
@app.get("/api/admin/panel")
def panel_admin(token: str = Depends(oauth2_scheme)):
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        rol = payload.get("rol")

        # Validación manual del rol
        if rol != "admin":
            raise HTTPException(
                status_code=status.HTTP_403_FORBIDDEN,
                detail="Acceso denegado: Se requiere rol de administrador"
            )
        return {"panel": "admin"}
    except JWTError:
        raise HTTPException(status_code=401, detail="Token inválido")
````

````python
%%writefile app_ia.py
from fastapi import FastAPI, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer
from pydantic import BaseModel
from jose import JWTError, jwt
import datetime

app = FastAPI(title="API Científico - Sesión 17")

SECRET_KEY = "mi_clave_secreta"
ALGORITHM = "HS256"
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="/api/token")

usuarios_db = {
    "admin": {"password": "admin123", "rol": "admin"},
    "ana": {"password": "secreta", "rol": "user"}
}

class LoginRequest(BaseModel):
    usuario: str
    password: str

# Dependencia para verificar rol
def verificar_admin(token: str = Depends(oauth2_scheme)):
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        if payload.get("rol") != "admin":
            raise HTTPException(
                status_code=status.HTTP_403_FORBIDDEN,
                detail="Acceso denegado: Se requiere rol de administrador"
            )
        return payload
    except JWTError:
        raise HTTPException(status_code=401, detail="Token inválido")

@app.post("/api/token")
def login(datos: LoginRequest):
    if datos.usuario in usuarios_db and usuarios_db[datos.usuario]["password"] == datos.password:
        rol = usuarios_db[datos.usuario]["rol"]
        expiracion = datetime.datetime.utcnow() + datetime.timedelta(minutes=30)
        token = jwt.encode({"sub": datos.usuario, "rol": rol, "exp": expiracion}, SECRET_KEY, algorithm=ALGORITHM)
        return {"access_token": token, "token_type": "bearer"}
    raise HTTPException(status_code=401, detail="Credenciales incorrectas")

@app.get("/api/perfil")
def perfil(token: str = Depends(oauth2_scheme)):
    try:
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        return {"usuario": payload.get("sub"), "rol": payload.get("rol")}
    except JWTError:
        raise HTTPException(status_code=401, detail="Token inválido")

@app.get("/api/admin/panel")
def panel_admin(payload: dict = Depends(verificar_admin)):
    return {"panel": "admin"}
````
