### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm

app = FastAPI(title="Tu OAuth2 Password Flow se verá así")

# Herramienta para leer el header Authorization
oauth2_scheme = OAuth2PasswordBearer(tokenUrl="token")

# Diccionario de usuarios
usuarios_db = {"ana": "secreta"}

@app.post("/token")
def login(form_data: OAuth2PasswordRequestForm = Depends()):
    # Validamos el usuario manualmente
    if form_data.username in usuarios_db and usuarios_db[form_data.username] == form_data.password:
        # Creamos un token simple (sin JWT, solo un string)
        token = f"token-de-{form_data.username}"
        return {"access_token": token, "token_type": "bearer"}

    raise HTTPException(status_code=401, detail="Credenciales incorrectas")

@app.get("/api/perfil")
def leer_perfil(token: str = Depends(oauth2_scheme)):
    # Validamos que el token tenga el formato correcto
    if token.startswith("token-de-"):
        usuario = token.replace("token-de-", "")
        return {"usuario": usuario}
    else:
        # Si el token es inválido, devolvemos 401 con el header WWW-Authenticate
        raise HTTPException(
            status_code=status.HTTP_401_UNAUTHORIZED,
            detail="Token inválido",
            headers={"WWW-Authenticate": "Bearer"},
        )
````

**CODIGO DE LA IA**

````python
%%writefile app_ia.py
from fastapi import FastAPI, Depends, HTTPException, status
from fastapi.security import OAuth2PasswordBearer, OAuth2PasswordRequestForm
from jose import JWTError, jwt

app = FastAPI(title="API Científico - Sesión 16")

SECRET_KEY = "mi_clave_secreta_super_segura"
ALGORITHM = "HS256"

oauth2_scheme = OAuth2PasswordBearer(tokenUrl="token")
usuarios_db = {"ana": "secreta"}

@app.post("/token")
def login(form_data: OAuth2PasswordRequestForm = Depends()):
    if form_data.username in usuarios_db and usuarios_db[form_data.username] == form_data.password:
        # Generamos un JWT real firmado
        token = jwt.encode({"sub": form_data.username}, SECRET_KEY, algorithm=ALGORITHM)
        return {"access_token": token, "token_type": "bearer"}
    raise HTTPException(status_code=401, detail="Credenciales incorrectas")

@app.get("/api/perfil")
def leer_perfil(token: str = Depends(oauth2_scheme)):
    try:
        # Decodificamos y validamos la firma del JWT
        payload = jwt.decode(token, SECRET_KEY, algorithms=[ALGORITHM])
        usuario = payload.get("sub")
        if usuario is None:
            raise HTTPException(status_code=401, detail="Token inválido")
        return {"usuario": usuario}
    except JWTError:
        # Si el token está alterado o vencido, lanzamos 401
        raise HTTPException(
            status_code=status.HTTP_401_UNAUTHORIZED,
            detail="Token inválido",
            headers={"WWW-Authenticate": "Bearer"},
        )
````
