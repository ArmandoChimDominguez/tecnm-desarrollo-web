### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, Depends, status
from pydantic import BaseModel

app = FastAPI(title="API Artesano - Sesión 12")

# 1. El modelo de entrada
class LibroRequest(BaseModel):
    titulo: str
    anio: int

# Capa de datos
class RepositorioMemoria:
    def __init__(self):
        self.libros = {}
        self.contador_id = 1

    def guardar(self, titulo: str, anio: int):
        libro = {"id": self.contador_id, "titulo": titulo, "anio": anio}
        self.libros[self.contador_id] = libro
        self.contador_id += 1
        return libro

    def obtener_todos(self):
        return list(self.libros.values())

# simula la BD
repositorio = RepositorioMemoria()

# Función de inyección de dependencia
def get_repositorio():
    return repositorio

# Endpoints que usan Depends
@app.get("/api/diagnostico")
def diagnostico(repo: RepositorioMemoria = Depends(get_repositorio)):
    return {"repositorio": repo.__class__.__name__}

@app.post("/api/libros", status_code=status.HTTP_201_CREATED)
def crear_libro(libro: LibroRequest, repo: RepositorioMemoria = Depends(get_repositorio)):
    # Validación manual
    if not libro.titulo.strip():
        raise HTTPException(status_code=422, detail="Título vacío")
    return repo.guardar(libro.titulo, libro.anio)

@app.get("/api/libros")
def listar_libros(repo: RepositorioMemoria = Depends(get_repositorio)):
    return {"libros": repo.obtener_todos()}
````

CODIGO DE LA IA

````python
%%writefile app_ia.py
from fastapi import FastAPI, Depends, status
from pydantic import BaseModel, Field
from abc import ABC, abstractmethod

app = FastAPI(title="API Científico - Sesión 12")

# 1. Modelo de entrada con validación
class LibroRequest(BaseModel):
    titulo: str = Field(..., min_length=1, description="Título del libro")
    anio: int

# 2. Interfaz del Repositorio (Contrato)
class IRepositorio(ABC):
    @abstractmethod
    def guardar(self, titulo: str, anio: int): ...

    @abstractmethod
    def obtener_todos(self): ...

# 3. Implementación concreta
class RepositorioMemoria(IRepositorio):
    def __init__(self):
        self._libros = {}
        self._contador_id = 1

    def guardar(self, titulo: str, anio: int):
        libro = {"id": self._contador_id, "titulo": titulo, "anio": anio}
        self._libros[self._contador_id] = libro
        self._contador_id += 1
        return libro

    def obtener_todos(self):
        return list(self._libros.values())

# 4. Instancia global
_repo_instancia = RepositorioMemoria()

# 5. Función de inyección
def get_repositorio() -> IRepositorio:
    return _repo_instancia

# 6. Endpoints
@app.get("/api/diagnostico")
def diagnostico(repo: IRepositorio = Depends(get_repositorio)):
    return {"repositorio": repo.__class__.__name__}

@app.post("/api/libros", status_code=status.HTTP_201_CREATED)
def crear_libro(libro: LibroRequest, repo: IRepositorio = Depends(get_repositorio)):
    # Pydantic ya validó el min_length=1, no hay if necesarios
    return repo.guardar(libro.titulo, libro.anio)

@app.get("/api/libros")
def listar_libros(repo: IRepositorio = Depends(get_repositorio)):
    return {"libros": repo.obtener_todos()}
````
