### 📄 `app.py`

Archivo principal de la implementación.

````python
%%writefile app.py
from fastapi import FastAPI, Request, Response, HTTPException
from pydantic import BaseModel
import uuid

app = FastAPI(title="API Artesano - Sesión 13")

class LoginRequest(BaseModel):
    usuario: str
    password: str

# Diccionario para guardar las sesiones
sesiones = {}

@app.post("/api/login")
def login(datos: LoginRequest, response: Response):
    # Validación manual del usuario
    if datos.usuario == "ana" and datos.password == "secreta":
        # Generamos un ID único
        id_sesion = str(uuid.uuid4())
        sesiones[id_sesion] = datos.usuario

        # Metemos la cookie en la respuesta
        response.set_cookie(
            key="session_id",
            value=id_sesion,
            httponly=True,
            samesite="lax"
        )
        return {"mensaje": "sesión iniciada"}

    raise HTTPException(status_code=401, detail="Datos incorrectos")

@app.get("/api/perfil")
def perfil(request: Request):
    # Buscamos la cookie que mandó el navegador
    id_sesion = request.cookies.get("session_id")

    # Si no hay cookie o no está en el diccionario, error
    if not id_sesion or id_sesion not in sesiones:
        raise HTTPException(status_code=401, detail="No autorizado")

    return {"usuario": sesiones[id_sesion]}

@app.post("/api/logout", status_code=204)
def logout(request: Request, response: Response):
    id_sesion = request.cookies.get("session_id")

    # Borramos la sesión de nuestro diccionario
    if id_sesion and id_sesion in sesiones:
        del sesiones[id_sesion]

    # Le decimos al navegador que borre la cookie
    response.delete_cookie("session_id")
    return
````

````python
%%writefile app_ia.py
from fastapi import FastAPI, Request, Response, HTTPException, status, Depends
from pydantic import BaseModel
import uuid

app = FastAPI(title="API Científico - Sesión 13")

class LoginRequest(BaseModel):
    usuario: str
    password: str

# Almacén de sesiones en memoria
sesiones_db = {}

@app.post("/api/login")
def login(datos: LoginRequest, response: Response):
    if datos.usuario != "ana" or datos.password != "secreta":
        raise HTTPException(status_code=401, detail="Credenciales incorrectas")

    # Generación del token de sesión
    session_id = str(uuid.uuid4())
    sesiones_db[session_id] = datos.usuario

    # Configuración de la cookie segura
    response.set_cookie(
        key="session_id",
        value=session_id,
        httponly=True,
        samesite="lax",
        secure=False # En producción debería ser True (requiere HTTPS)
    )
    return {"mensaje": "sesión iniciada"}

@app.get("/api/perfil")
def perfil(request: Request):
    session_id = request.cookies.get("session_id")
    usuario = sesiones_db.get(session_id)

    if not usuario:
        raise HTTPException(status_code=status.HTTP_401_UNAUTHORIZED, detail="No autorizado")

    return {"usuario": usuario}

@app.post("/api/logout", status_code=status.HTTP_204_NO_CONTENT)
def logout(request: Request, response: Response):
    session_id = request.cookies.get("session_id")

    if session_id in sesiones_db:
        del sesiones_db[session_id]

    response.delete_cookie("session_id")
````
